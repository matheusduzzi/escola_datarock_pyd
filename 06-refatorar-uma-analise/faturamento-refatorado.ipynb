# Faturamento do semestre, versão em módulo

As regras moram em `faturamento.py`. Aqui eu só chamo as funções e exploro.

```python
from pathlib import Path

import faturamento

PASTA_DADOS = Path("../dados/loja")
```

```python
pedidos, produtos = faturamento.ler_base(PASTA_DADOS)
base = faturamento.entregues_com_categoria(pedidos, produtos)
len(base)
```

```python
semestre = faturamento.por_mes_e_categoria(faturamento.no_periodo(base, "2026-01", "2026-06"))
semestre
```

```python
# explorar sem mexer em base: o resultado ganha outro nome
trimestre = faturamento.por_mes_e_categoria(faturamento.no_periodo(base, "2026-04", "2026-06"))
print("Total do semestre:", round(semestre.sum().sum(), 2))
print("Total do 2º trimestre:", round(trimestre.sum().sum(), 2))
```
