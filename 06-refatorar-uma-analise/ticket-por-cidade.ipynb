# Ticket médio por cidade

Pedidos entregues. Ticket = média do `valor` dos pedidos de cada cidade (pedido sem valor anotado fica fora da média). A gerente quer as 3 cidades com o maior ticket.

```python
import pandas as pd

pedidos = pd.read_csv("../dados/loja/pedidos.csv")
clientes = pd.read_csv("../dados/loja/clientes.csv")
```

```python
entregues = pedidos[pedidos["status"] == "entregue"]
base = entregues.merge(clientes[["id_cliente", "cidade"]], on="id_cliente")
base.shape
```

```python
por_cidade = resumo(base)
por_cidade
```

```python
# só para ver Campinas
base = base[base["cidade"] == "Campinas"]
base["valor"].describe().round(2)
```

```python
def resumo(df):
    tabela = df.groupby("cidade").agg(pedidos=("id_pedido", "count"), ticket=("valor", "mean"))
    return tabela.round(2).sort_values("ticket", ascending=False)
```

```python
maiores = resumo(base).head(3)
print(maiores)
```
